Project name : Employee Data Analysis

Step 1 :- loading library

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Step 2 :- Filtering warning

```python
import warnings
warnings.filterwarnings("ignore")
```

Step 3 :- loading data sets

```python
df1 = pd.read_excel('/content/2020.xlsx')
```

```python
df2 = pd.read_excel('/content/2021.xlsx')
```

```python
df3 = pd.read_excel('/content/2022.xlsx')
```

```python
# combining together

df = pd.concat([df1,df2,df3],ignore_index=True)
```

```python
# fetching combined data

df
```

Step 4 :- Data Exploration

```python
# displays basic info about data

print("Basic Information about data :\n")
df.info()

print("\nTotal records :",df.size)

print("\n Number of rows :", df.shape[0])

print("\n Number of columns :",df.shape[1])

print("\nDescribing numeric columns :\n", df.describe())
```

```python
# fetching first 10 records

df.head(10)
```

```python
# fetching last 10 records

df.tail(10)
```

Step 5 :- Data Cleaning

```python
print("Checking with null values :\n",df.isnull().sum())
```

```python
print("checking with duplicates :\n",df.duplicated().sum())
```

Step 6 :- Data Analysis

```python
# showing department & position wise total no of employee

df.groupby(["Department","Position"]).agg(Total_Employee=('Employee ID', 'count'))
```

```python
# showing employee count - gender wise

df.groupby("Gender").agg(Total_Employee=('Employee ID', 'count'))
```

```python
# showing employee total - gender & martial status

df.groupby(["Gender","Marital Status"]).agg(Total_Employee=('Employee ID', 'count'))
```

```python
# showing department & position wise highest salary employee

df.groupby(["Department","Position"]).agg(Employee_Name=('Name','unique'),Highest_salary=('Salary', 'max'))
```

```python
# showing total employee count - experience years & gender

df.groupby(["Gender","Experience Years"]).agg(Total_Employee=('Employee ID', 'count'))
```

```python
# showing country wise total no of employee

df.groupby(["Country","Nationality",]).agg(Total_Employee=('Employee ID', 'count'))
```

Step 7 :- Data Visualization

```python
# showing department  wise total no of employee

Totals = df.groupby("Department").agg(Total_Employee=('Employee ID', 'count'))

# plotting graph

plt.figure(figsize=(10,8))

sns.barplot(data=Totals,x=Totals.index,y='Total_Employee')

plt.title("Departmentwise Total Employee Count")
plt.xlabel("Department Name")
plt.ylabel("Total Employee Count")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# showing country wise total no of employee

X=df.groupby("Country").agg(Total_Employee=('Employee ID', 'count'))

# plotting graph

plt.figure(figsize=(12,8))

plt.pie(X['Total_Employee'],labels=X.index,autopct='%1.2f%%')

plt.title("Total No of Employee In Country ")

plt.show()
```

```python
sns.histplot(data=df,x="Salary",bins=20,kde=True)

plt.title("Distribution of salary ")

plt.xticks(rotation=45)
plt.show()
```

```python
# plotting heatmap to show correlation between numeric columns

sns.heatmap(df.corr(numeric_only=True),annot=True)

plt.show()
```
